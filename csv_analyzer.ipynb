## 📋 Overview
This notebook demonstrates how to use **LangChain** to analyze CSV files using natural language queries.

### What You'll Learn:

- ✅ Ask questions in plain English

### Technologies Used:

- **Python** - Programming language

---
## Step 1: Import Required Libraries

```python
import os
import pandas as pd
from dotenv import load_dotenv
from langchain_experimental.agents import create_pandas_dataframe_agent
from langchain_openai import ChatOpenAI

print("✅ Libraries imported successfully!")
```

---
## Step 2: Load Environment Variables
This loads your OpenAI API key from the `.env` file.

```python
# Load environment variables
load_dotenv()

# Verify API key is loaded
api_key = os.getenv("OPENAI_API_KEY")
if api_key:
    print("✅ OpenAI API Key loaded successfully!")
    print(f"   Key starts with: {api_key[:10]}...")
else:
    print("❌ ERROR: OpenAI API Key not found in .env file")
```

---
## Step 3: Create Sample CSV Data
For this demo, we'll create a sample sales dataset. You can replace this with your own CSV file later.

```python
# Create sample sales data
sample_data = {
    'Date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05',
             '2024-01-06', '2024-01-07', '2024-01-08', '2024-01-09', '2024-01-10'],
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop',
                'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Headphones'],
    'Category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Electronics',
                 'Accessories', 'Accessories', 'Electronics', 'Electronics', 'Accessories'],
    'Quantity': [2, 5, 3, 1, 1, 10, 4, 2, 3, 6],
    'Price': [50000, 500, 1500, 15000, 50000, 500, 1500, 15000, 50000, 2000],
    'Customer': ['John', 'Alice', 'Bob', 'Charlie', 'David',
                 'Eve', 'Frank', 'Grace', 'Henry', 'Ivy'],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Mumbai', 'Delhi',
             'Bangalore', 'Mumbai', 'Delhi', 'Bangalore', 'Mumbai']
}

# Create DataFrame
df = pd.DataFrame(sample_data)

# Calculate total sales
df['Total'] = df['Quantity'] * df['Price']

# Save to CSV
csv_path = 'sample_sales.csv'
df.to_csv(csv_path, index=False)

print("✅ Sample CSV created successfully!")
print(f"\n📊 Preview of the data:")
print(df.head())
print(f"\n📈 Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
```

---
## Step 4: Load Your CSV File
You can replace `sample_sales.csv` with your own CSV file path.

```python
# Load CSV file
csv_file = 'sample_sales.csv'  # Change this to your CSV file path

try:
    df = pd.read_csv(csv_file)
    print(f"✅ CSV file loaded successfully!")
    print(f"\n📊 Dataset Info:")
    print(f"   - Rows: {df.shape[0]}")
    print(f"   - Columns: {df.shape[1]}")
    print(f"\n📋 Column names:")
    print(f"   {', '.join(df.columns.tolist())}")
    print(f"\n👀 First 5 rows:")
    display(df.head())
except FileNotFoundError:
    print(f"❌ ERROR: File '{csv_file}' not found!")
except Exception as e:
    print(f"❌ ERROR: {e}")
```

---
## Step 5: Initialize the LangChain Agent
This creates an AI agent that can analyze your CSV data.

```python
# Initialize OpenAI model
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,  # More deterministic responses
    api_key=os.getenv("OPENAI_API_KEY")
)

# Create Pandas DataFrame agent
agent = create_pandas_dataframe_agent(
    llm=llm,
    df=df,
    verbose=True,  # Shows the agent's thinking process
    allow_dangerous_code=True,  # Required for pandas operations
    agent_type="openai-tools"
)

print("✅ LangChain CSV Agent created successfully!")
print("\n🤖 Agent is ready to answer your questions!")
```

---
## Step 6: Ask Questions About Your Data
Now you can ask questions in natural language!

### Example 1: Basic Statistics

```python
question = "How many rows are in the dataset?"

print(f"❓ Question: {question}")
print("\n🤖 Agent Response:")
print("=" * 60)

response = agent.invoke(question)
print(f"\n✅ Answer: {response['output']}")
```

## 🎯 How It Works Behind the Scenes

1. **You ask a question** in natural language
2. **LangChain agent** understands your intent
3. **Agent generates Python/Pandas code** to query the data
4. **Code executes** on your DataFrame
5. **Results are formatted** and returned in natural language
